## 1. Model

*(Replace the equations below with your own. Use `$...$` for inline math and `$$...$$` for display math.)*

The baseline specification is:

$$
\log w_{it} = \alpha + \beta_1 \, \text{Signal}_{it} + \beta_2 X_{it} + \mu_i + \epsilon_{it}
$$

where $w_{it}$ is the wage of individual $i$ at time $t$, $\text{Signal}_{it}$ is the signal of interest, $X_{it}$ is a vector of controls, and $\mu_i$ captures individual fixed effects.

The identifying assumption is:

$$
\mathbb{E}[\epsilon_{it} \mid \text{Signal}_{it},\, X_{it},\, \mu_i] = 0
$$
